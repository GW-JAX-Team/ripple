```python
import jax
import jax.numpy as jnp

import lal
import lalsimulation as lalsim
from lalsimulation import SimIMRPhenomXPSpinTaylorAngles

from ripplegw.constants import MSUN
from ripplegw.waveforms.LALSimIMRPhenomX import XLALSimIMRPhenomXPSpinTaylorAngles 

jax.config.update("jax_enable_x64", True)

import matplotlib.pyplot as plt
```

```python
sampling_frequency = 256

freqs = jnp.linspace(20.0, sampling_frequency / 2.0, num=sampling_frequency)

m1_si = 30.0 * MSUN
m2_si = 20.0 * MSUN
chi1x = 0.8
chi1y = 0.2
chi1z = 0.4
chi2x = 0.1
chi2y = 0.2
chi2z = 0.3
deltaF = freqs[1] - freqs[0]
fRef = 20.0
mprime = 2
lal_params = {
    "IMRPhenomXPrecVersion": 330, 
    "PNRUseTunedAngles": 0,
    "AntisymmetricWaveform": 0,
    "PNRUseTunedCoprec": 0,
    "ExpansionOrder": 5,
}

print("Computing SpinTaylor angles with ripple...")

alphas_ripple, gammas_ripple, cosbetas_ripple = XLALSimIMRPhenomXPSpinTaylorAngles(
    m1_si,
    m2_si,
    chi1x,
    chi1y,
    chi1z,
    chi2x,
    chi2y,
    chi2z,
    freqs[0],
    freqs[-1],
    deltaF,
    fRef,
    0.0,
    lal_params
)

laldict = lal.CreateDict()
lalsim.SimInspiralWaveformParamsInsertPhenomXPrecVersion(laldict, 330)

print("Computing SpinTaylor angles with LAL...")

alphas_lal, gammas_lal, cosbetas_lal = SimIMRPhenomXPSpinTaylorAngles(
    m1_si,
    m2_si,
    chi1x,
    chi1y,
    chi1z,
    chi2x,
    chi2y,
    chi2z,
    deltaF,
    fRef,
    0.0,
    lal_params
)

print("#" * 70)
print("ripple SpinTaylor angles")
print(f"alphas: {alphas_ripple}\ngammas: {gammas_ripple}\ncosbetas: {cosbetas_ripple}")
print("#" * 70)
print("lal SpinTaylor angles")
print(f"alphas: {alphas_lal.data}\ngammas: {gammas_lal.data}\ncosbetas: {cosbetas_lal.data}")
print("#" * 70)
```
